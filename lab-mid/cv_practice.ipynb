```python
import cv2
import matplotlib.pyplot as plt
```

```python
image_path = '/content/dahlia-1508785047.jpg.avif'
image = cv2.imread(image_path)
```

```python
# rgb image show

if image is None:
  print("error no image")
else:
  image_rgb = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
  plt.figure(figsize=(8,6))
  plt.axis('off')
  plt.title("flower", fontsize=20,color='pink',pad=15)
  plt.imshow(image_rgb)
  plt.show()

```

```python
#gray image

image = cv2.imread(image_path)

gray_image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
plt.figure(figsize=(6,6))
plt.axis('off')
plt.title("flower", fontsize=20,color='pink',pad=15)
plt.imshow(gray_image,cmap='gray')
plt.show()
```

```python
#resize image

image = cv2.imread(image_path)
image_rgb= cv2.cvtColor(image , cv2.COLOR_BGR2RGB)

scale_fact = 0.3
resized_image = cv2.resize(image_rgb,None , fx=scale_fact ,fy=scale_fact)

new_height,new_width = resized_image.shape[:2]
ori_height,ori_width = image_rgb.shape[:2]

print(f"original res = {ori_width}x{ori_height}")
print(f"new res= {new_width}x{new_height}")



plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title(
        "Original ",
        fontsize=16,
        color="darkblue"
    )

    # Blurred ROI
plt.subplot(1, 2, 2)
plt.imshow(resized_image)
plt.title(
"reszed",
           fontsize=16,
        color="darkred"
    )

plt.show()
```

```python
#blur

image = cv2.imread(image_path)

image_rgb = cv2.cvtColor(image , cv2.COLOR_BGR2RGB)

blur_image = cv2.GaussianBlur(image_rgb , (41,31), 0)

plt.imshow(blur_image)
plt.title("blur image", color='blue', fontsize=10)
plt.show()
```

```python
#croppimg

image=cv2.imread(image_path)

image_rgb = cv2.cvtColor(image , cv2.COLOR_BGR2RGB)

height,width = image_rgb.shape[:2]

mid_x = width // 2

cropped_img_left = image_rgb[:,:mid_x]
cropped_img_right = image_rgb[:,mid_x:]

plt.subplot(1,3,1)
plt.imshow(cropped_img_right)
plt.title("croopped right")

plt.subplot(1,3,2)
plt.imshow(cropped_img_left)
plt.title("left crop")

plt.subplot(1,3,3)
plt.imshow(image_rgb)
plt.title("original")


plt.show()
```

```python
#text on img

image=cv2.imread(image_path)

image_rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)

text = "FLOWERRR"
font=cv2.FONT_HERSHEY_SIMPLEX

cv2.putText(image_rgb, text,(110,110),font,5,(255,60,100),2)

plt.imshow(image_rgb)
plt.show()
```

```python
# analyse image using pandas
import pandas as pd
from IPython.display import display

image = cv2.imread(image_path)

flat_img = image.reshape(-1,3)
img_df = pd.DataFrame(flat_img, columns=['B','G','R'])

print('basic staticts')
display(img_df.describe())
```

```python
# image thresholdin

image = cv2.imread(image_path)

gray_img = cv2.cvtColor(image , cv2.COLOR_BGR2GRAY)

ret, thres_img = cv2.threshold(gray_img , 100,200, cv2.THRESH_BINARY)

plt.imshow(thres_img , cmap='gray')
plt.show()
```

```python
# rotation


image=cv2.imread(image_path)

image_rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)

height,width = image_rgb.shape[:2]

center = (width/2 , height/2)
rot_matrix = cv2.getRotationMatrix2D(center , 160,1)
rot_img = cv2.warpAffine(image_rgb , rot_matrix,(width,height))

plt.imshow(rot_img)
plt.show()
```

```python
# blendin

image_path2 = '/content/gettyimages-2165950545-69600af8c9b0d.jpg.avif'

image1 = cv2.imread(image_path)
image2 = cv2.imread(image_path2)

image1_rgb=cv2.cvtColor(image1,cv2.COLOR_BGR2RGB)
image2_rgb=cv2.cvtColor(image2,cv2.COLOR_BGR2RGB)




target_height, target_width = image2_rgb.shape[:2]
image1_resized = cv2.resize(image1_rgb , (target_width,target_height))

blend_img = cv2.add(image1_resized , image2_rgb)

plt.imshow(blend_img)
plt.show()
```
